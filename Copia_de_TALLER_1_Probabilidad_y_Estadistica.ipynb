```python
# ACTIVIDAD 1.1 — Parte 2: Muestras aleatorias de tamaño 50 con semilla reproducible
# Gráfico de barras que muestra la distribución de niveles educativos en las dos muestras
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Cargar el archivo CSV
df = pd.read_csv('/content/sgdata.csv')  # Asegúrate de que el nombre del archivo sea correcto

# Semilla asignada
seed = 380
rng = np.random.default_rng(seed)

# Crear las dos muestras de tamaño 50 desde la misma semilla
m1 = df.sample(n=50, random_state=rng.integers(0, 1_000_000))
m2 = df.sample(n=50, random_state=rng.integers(0, 1_000_000))

# ---------- Gráfico para Muestra 1 ----------
m1_edu_counts = m1['Education'].value_counts()

plt.figure(figsize=(8, 5))
m1_edu_counts.plot(kind='bar', color='lightgreen')
plt.title('Distribución de niveles educativos - Muestra 1 (50 personas)')
plt.xlabel('Nivel educativo')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# ---------- Gráfico para Muestra 2 ----------
m2_edu_counts = m2['Education'].value_counts()

plt.figure(figsize=(8, 5))
m2_edu_counts.plot(kind='bar', color='orange')
plt.title('Distribución de niveles educativos - Muestra 2 (50 personas)')
plt.xlabel('Nivel educativo')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

**¿Qué nivel educativo es predominante?**

En ambas muestras, así como en la población total observamos que en los gráficos, que el nivel educativo más común es el de “high school” (secundaria), ya que en ambas muestras son más de 35 personas las cuales se encuentran en este grupo. Esto respalda lo que ya habíamos visto en la población total, donde también hubo una gran mayoría en este nivel. La similitud entre nuestra muestra y la población total nos dice que esta muestra es representativa del total, lo que reafrima que la mayoría solo cursa hasta secundaria.

**¿Qué razones socioeconómicas podrían explicar esta distribución?**
 En este caso las razones serian las mismas que afectan a la poblacion total, tales como limitaciones financieras,acceso geografico y calidad en la educación.

```python
# Punto 1.2 - Gráfico de sectores (pie) poblacion total
# Este gráfico representa la proporción de individuos según el tamaño del asentamiento (Settlement size)
# 0 = Área rural o asentamiento pequeño
# 1 = Ciudad mediana o área semiurbana
# 2 = Ciudad grande o área metropolitana

import matplotlib.pyplot as plt

# Crear una copia del DataFrame original para no modificar el original si se vuelve a usar
df_pie = df.copy()

# Reemplazar los valores numéricos por etiquetas descriptivas directamente en una copia
df_etiquetado = df.copy()
df_etiquetado['Settlement size'] = df_etiquetado['Settlement size'].astype(str).replace({
    '0': 'Área rural / Asent. pequeño',
    '1': 'Ciudad mediana / Semiurbana',
    '2': 'Ciudad grande / Metropolitana'
})

# Contar la frecuencia de cada categoría
settlement_counts = df_etiquetado['Settlement size'].value_counts()

# Crear gráfico de sectores con nombres
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.pie(settlement_counts, labels=settlement_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribución por tamaño del asentamiento (Settlement size)')
plt.axis('equal')
plt.show()
```

Como resultado, la población que vive en zonas rurales tiene menos oportunidades de crecimiento económico en comparación con los residentes de la ciudad. Por lo tanto, la ubicación de la residencia puede determinar la igualdad económica.

```python
# Punto 1.2  grafico sectores pie para muestras m1 y m2
seed = 380
rng = np.random.default_rng(seed)

# Muestras de tamaño 50 usando la misma semilla
m1 = df.sample(n=50, random_state=rng.integers(0, 1_000_000))
m2 = df.sample(n=50, random_state=rng.integers(0, 1_000_000))

# Etiquetar los valores de Settlement size
etiquetas = {
    0: 'Área rural / Asent. pequeño',
    1: 'Ciudad mediana / Semiurbana',
    2: 'Ciudad grande / Metropolitana'
}

# Reemplazar en m1 y m2
m1_etiquetado = m1.copy()
m1_etiquetado['Settlement size'] = m1_etiquetado['Settlement size'].replace(etiquetas)

m2_etiquetado = m2.copy()
m2_etiquetado['Settlement size'] = m2_etiquetado['Settlement size'].replace(etiquetas)

# Contar frecuencias
settlement_m1 = m1_etiquetado['Settlement size'].value_counts()
settlement_m2 = m2_etiquetado['Settlement size'].value_counts()

# -------- Gráfico de sectores para Muestra 1 --------
plt.figure(figsize=(8, 6))
plt.pie(settlement_m1, labels=settlement_m1.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribución por tamaño del asentamiento - Muestra 1')
plt.axis('equal')
plt.show()

# -------- Gráfico de sectores para Muestra 2 --------
plt.figure(figsize=(8, 6))
plt.pie(settlement_m2, labels=settlement_m2.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribución por tamaño del asentamiento - Muestra 2')
plt.axis('equal')
plt.show()
```

```python
# Punto 1.3 Generen un histograma de los ingresos.
#¿La distribución es simétrica o asimétrica?
#¿Qué nos indica esto sobre la desigualdad de ingresos?

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.hist(df['Income'], bins=30, color='skyblue', edgecolor='black')
plt.title('Histograma de Ingresos')
plt.xlabel('Ingreso')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.tight_layout()
plt.show()
```

**¿La distribución es simétrica o asimétrica?**
La distribución es asimétrica hacia la derecha (o positivamente asimétrica). Esto significa que la mayoría de los ingresos se agrupan en la parte baja y media de la distribución, alrededor de 100.000, mientras que hay una larga cola hacia la derecha, es decir, pocas personas tienen ingresos muy altos.

**¿Qué nos indica esto sobre la desigualdad de ingresos?**
Lo que muestra es que hay demasiada desigualdad en los ingresos. La mayoría de la gente gana más o menos lo mismo, en un rango medio, pero hay unos pocos que ganan muchísimo más. Entonces la gráfica se ve “apelmazada” hacia la izquierda, con una cola muy larga a la derecha. Es decir, aunque la mayoría gana similar, hay unos pocos que ganan sueldos tan altos que hacen subir el promedio, aunque no son la generalidad ni representan lo que gana la mayoría.

```python
# Histograma de ingresos - Muestra 1
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Semilla reproducible
seed = 380
rng = np.random.default_rng(seed)

# Dos muestras aleatorias distintas desde la misma semilla
m1 = df.sample(n=50, random_state=rng.integers(0, 1_000_000))
m2 = df.sample(n=50, random_state=rng.integers(0, 1_000_000))

# -------- Histograma de ingresos - Muestra 1 --------
plt.figure(figsize=(8, 5))
plt.hist(m1['Income'], bins=20, color='skyblue', edgecolor='black')
plt.title('Histograma de Ingresos - Muestra 1 (50 personas)')
plt.xlabel('Ingreso')
plt.ylabel('Frecuencia')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# -------- Histograma de ingresos - Muestra 2 --------
plt.figure(figsize=(8, 5))
plt.hist(m2['Income'], bins=20, color='salmon', edgecolor='black')
plt.title('Histograma de Ingresos - Muestra 2 (50 personas)')
plt.xlabel('Ingreso')
plt.ylabel('Frecuencia')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

```

**¿La distribución es simétrica o asimétrica?**
En ambas muestras se evidencia que los ingresos tienen una distribución asimétrica.En el histograma de la muestra 1, la mayoría de las personas se encuentran en un rango de ingresos intermedios, pero hay algunas que ganan mucho más, lo que hace que la "cola" del gráfico se alargue hacia la derecha. A esto lo llamamos asimetría positiva o sesgo a la derecha. Lo mismo pasa en la muestra 2; aquí se ven un poco más esparcidos, con la mayoria de personas ganando menos y unos pocos con altos ingresos que empujan la gráfica hacia ese lado.
**¿Qué nos indica esto sobre la desigualdad de ingresos?**
Esto nos muestra que no todos ganan lo mismo y que hay desigualdad. En otras palabras, aunque hay muchas personas que tienen ingresos similares, también hay unas pocas que ganan muchísimo más que los demás. Esto provoca un desequilibrio: el promedio se eleva por esos pocos que ganan más, pero en realidad la mayoría está por debajo de esa cifra promedio

```python
# PUNTO 4 - DESIGUALDAD DE CHEBYSHEV (POBLACIÓN TOTAL)

# Media y desviación estándar ya conocidas
media = df['Income'].mean()
desviacion = df['Income'].std()

# 4.1 - Porcentaje mínimo según la desigualdad de Chebyshev para k = 2
k = 2
chebyshev_minimo = 1 - (1 / k**2)
porcentaje_chebyshev = chebyshev_minimo * 100

print(f"PUNTO 4.1:")
print(f"Según la desigualdad de Chebyshev, al menos el {porcentaje_chebyshev:.2f}% de los datos estarán dentro de 2 desviaciones estándar de la media.")

# 4.2 - Porcentaje real observado en la población total
lim_inferior = media - k * desviacion
lim_superior = media + k * desviacion

# Datos dentro del rango
dentro_rango = df[(df['Income'] >= lim_inferior) & (df['Income'] <= lim_superior)]
porcentaje_real = (len(dentro_rango) / len(df)) * 100

print(f"\nPUNTO 4.2:")
print(f"Porcentaje real de datos dentro del rango: {porcentaje_real:.2f}%")

# Comparación visual (opcional)
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(['Mínimo según Chebyshev', 'Porcentaje real'], [porcentaje_chebyshev, porcentaje_real], color=['skyblue', 'lightgreen'])
plt.ylim(0, 100)
plt.ylabel('Porcentaje (%)')
plt.title('Comparación: Chebyshev vs Porcentaje Real (2 desv. estándar)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

Como podemos observar los datos reales son mayores que la desigualdad de chebyshev dando una diferencia de casi 20%.
Esto nos demuestra que Chebyshev nos ayuda a encontrar la desviación estandar pero con un resultado menor al resultado real.


```python
# 4.1 Utilizando la desviación estándar calculada, determinen qué porcentaje mínimo de datos se espera dentro de 2 desviaciones estándar según la desigualdad de Chebyshev.
import pandas as pd
import numpy as np

# Cargar tu DataFrame (ajusta esta parte si ya tienes df definido)
# df = pd.read_csv('tu_archivo.csv')  # o como sea que cargues tus datos

# -------------------------------
# SEMILLA ÚNICA Y MUESTRAS
# -------------------------------
rng = np.random.default_rng(seed=380)

# Usamos el mismo generador para ambas muestras
muestra1 = df.sample(n=50, random_state=rng.integers(0, 1000000))
muestra2 = df.sample(n=50, random_state=rng.integers(0, 1000000))

# -------------------------------
# FUNCIÓN CHEBYSHEV
# -------------------------------
def chebyshev_info(nombre_muestra, df_muestra):
    media = df_muestra['Income'].mean()
    std_dev = df_muestra['Income'].std()

    # Intervalo de 2 desviaciones estándar
    lower_bound = media - 2 * std_dev
    upper_bound = media + 2 * std_dev

    # Porcentaje real de datos dentro del intervalo
    dentro_intervalo = df_muestra[(df_muestra['Income'] >= lower_bound) & (df_muestra['Income'] <= upper_bound)]
    porcentaje_real = (len(dentro_intervalo) / len(df_muestra)) * 100

    # Porcentaje mínimo esperado según Chebyshev (k = 2)
    k = 2
    porcentaje_chebyshev = (1 - 1 / k**2) * 100

    print(f"--- {nombre_muestra} ---")
    print(f"Media: {media:.2f}")
    print(f"Desviación estándar: {std_dev:.2f}")
    print(f"Intervalo (media ± 2σ): [{lower_bound:.2f}, {upper_bound:.2f}]")
    print(f"Porcentaje mínimo esperado por Chebyshev: {porcentaje_chebyshev:.2f}%")
    print(f"Porcentaje real observado: {porcentaje_real:.2f}%")
    print()

# -------------------------------
# APLICAR A LAS DOS MUESTRAS
# -------------------------------
chebyshev_info("Muestra 1", muestra1)
chebyshev_info("Muestra 2", muestra2)
```
