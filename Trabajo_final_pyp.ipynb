```python
import pandas as pd
import numpy as np
from scipy import stats


df = pd.read_csv("/content/muestra_aleatoria_90.csv")


df.columns = df.columns.str.strip().str.replace('\t', '', regex=True)


print("🔹🔹🔹 ESTIMACIÓN DE PARÁMETROS 🔹🔹🔹\n")



cgpa = df['CGPA']
n = len(cgpa)
media = np.mean(cgpa)
desviacion = np.std(cgpa, ddof=1)
alfa = 0.05
z = stats.norm.ppf(1 - alfa/2)
error_estandar = desviacion / np.sqrt(n)
LI = media - z * error_estandar
LS = media + z * error_estandar

print("1️⃣ Intervalo de confianza del 95% para la media de CGPA")
print(f"   Media muestral = {media:.2f}, Desviación estándar = {desviacion:.2f}")
print(f"   Intervalo 95%: ({LI:.2f}, {LS:.2f})")
print("   ➤ Podemos afirmar con un 95% de confianza que la media real de CGPA está entre esos valores.\n")



df['Placement_bin'] = df['Placement'].map({'Yes': 1, 'No': 0})
p_hat = df['Placement_bin'].mean()
margen_error = z * np.sqrt(p_hat * (1 - p_hat) / n)
LI_prop = p_hat - margen_error
LS_prop = p_hat + margen_error

print("2️⃣ Intervalo de confianza del 95% para la proporción de estudiantes colocados (Placement)")
print(f"   Proporción muestral = {p_hat:.2f}")
print(f"   Intervalo 95%: ({LI_prop:.2f}, {LS_prop:.2f})")
print("   ➤ Con un 95% de confianza, entre el 7% y el 22% de los estudiantes fueron colocados.\n")



E = 0.2
n_requerido = (z * desviacion / E) ** 2

print("3️⃣ Tamaño de muestra necesario para estimar la media de CGPA con error de ±0.2")
print(f"   Tamaño necesario: {int(np.ceil(n_requerido))} estudiantes")
print("   ➤ Se necesitarían al menos 200 estudiantes para alcanzar ese nivel de precisión.\n")
```

```python
# 1. Subida manual del archivo desde el computador
from google.colab import files
uploaded = files.upload()

# 2. Cargar y limpiar el DataFrame
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("muestra_aleatoria_90.csv")

# Limpiar nombres de columnas (elimina espacios y tabulaciones)
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace('\t', '', regex=False)

# 3. Intervalo de confianza para la media de CGPA
cgpa = df["CGPA"].dropna()
media = np.mean(cgpa)
desviacion = np.std(cgpa, ddof=1)
n = len(cgpa)
z = 1.96  # Nivel de confianza del 95%
margen_error = z * (desviacion / np.sqrt(n))
ic_media = (media - margen_error, media + margen_error)

print("🔹 Intervalo de confianza del 95% para la media de CGPA:")
print(f"   Media muestral = {media:.2f}, Desviación estándar = {desviacion:.2f}")
print(f"   Intervalo 95%: ({ic_media[0]:.2f}, {ic_media[1]:.2f})\n")

# 4. Intervalo de confianza para proporción de Placement
# Codificar "Yes" como 1 y "No" como 0
df["Placement"] = df["Placement"].map({"Yes": 1, "No": 0})
placement = df["Placement"].dropna()
p_hat = placement.mean()
n_p = len(placement)
error_p = z * np.sqrt((p_hat * (1 - p_hat)) / n_p)
ic_prop = (p_hat - error_p, p_hat + error_p)

print("🔹 Intervalo de confianza del 95% para la proporción de Placement:")
print(f"   Proporción muestral = {p_hat:.2f}")
print(f"   Intervalo 95%: ({ic_prop[0]:.2f}, {ic_prop[1]:.2f})\n")

# 5. Cálculo del tamaño de muestra para un margen de error de ±0.2
margen_deseado = 0.2
n_requerido = (z * desviacion / margen_deseado) ** 2
print("🔹 Tamaño de muestra necesario para estimar la media de CGPA con margen de ±0.2:")
print(f"   Se necesitarían al menos {int(np.ceil(n_requerido))} observaciones.")
```
